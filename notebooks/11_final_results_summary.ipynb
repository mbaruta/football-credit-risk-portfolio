## 5. Interpretation to preserve in the final portfolio

The final narrative should remain consistent with the frozen methodology and results:

- **V1 was deliberately simple**, not an incorrect model.
- The simple financial snapshot in V1 was weak relative to the persistence baseline.
- V2 substantially improved the financial specification by adding current loss status, previous loss history, and exact one-year financial changes.
- The improvement from financial V1 to financial V2 is meaningful, but the **increment over the persistence baseline remains small**.
- Sporting variables did **not demonstrate robust incremental predictive value** beyond the V2 financial specification in this sample and period.
- This does **not** justify the causal statement that sporting performance “does not matter”.
- The restricted Gradient Boosting comparator did not outperform Ridge, so the main limitation cannot be attributed simply to a linear functional form.
- The `without_2020` analysis is a **sensitivity analysis**, not a replacement for the main specification.
- The project should not continue with broad model shopping merely to obtain a higher ROC-AUC.
